```python
# Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import datetime
```

```python
# Constant values
S_0 = 1367 # [W/m2] Solar constant; perpendicular to the top of the atmosphere
```

```python
# Settings; time and location for the measurements
lat = 50
lon = 20
lon_0 = 15
day_no = 13
t = 12
```

```python
### top of the atmosphere ###
```

```python
# Solar constant changes, as the Earths distance to the Sun changes
days = np.arange(1, 366)


S_0d = S_0 * (1 + 0.033 * np.cos(2 * np.pi * days / 365)) # Solar constant calcluated for every day of the year

S_0d_values = pd.DataFrame( 
    {'Day number' : days,
    'S_0d': S_0d_values}
)
```

```python
# Plotting S_0d
plt.plot(S_0d_values['Day number'], S_0d_values['S_0d'])
plt.xlabel('Day of the year')
plt.ylabel('Solar radiation hitting top of the atosmphere [W/m2]')
```

```python
# Time correction for the unevenness of the apparent motion of the Sun across the celestial sphere [min]
# I don't understand the physics behing this, what is this time correction delta?
N_d_delta = (days - 81) / 364
delta = 9.87 * np.sin(4 * np.pi * N_d_delta) - 7.53 * np.cos(2 * np.pi * N_d_delta) - 1.5 * np.sin(2 * np.pi * N_d_delta) 
plt.plot(delta)
```

```python
# Local sun time
hours = np.arange(12)

for hour in hours:
    t_local = hour + (4 * (lon - lon_0) + delta) / 60 # [h]
t_local
```

```python
# tau - Hourly angle of the sun
tau = (t_local - 12) * 2 * np.pi / 24
```

```python
# Sun declination
N_d_declination = 2 * np.pi * days / 366
declination_degrees = 0.33281 - 22.984 * np.cos(N_d_declination) + 3.7872 * np.sin(N_d_declination) 
- 0.3499 * np.cos(2 * N_d_declination)  + 0.03205 * np.sin(2 * N_d_declination) 
- 0.1398 * (3 * N_d_declination) + 0.07187 * (N_d_declination) # [degrees] 

declination_radians = 2 * np.pi * declination_degrees / 366 # [rad] 
```

```python
# Sun position calculation
h = np.arcsin(np.sin(declination_radians) * np.sin(lon) + np.cos(declination_radians) * np.cos(tau) * np.cos(lon)) # h is the angular height of the sun above the horizon
I_0h = S_0d * np.sin(h)
plt.plot(I_0h)
plt.xlabel('Day number')
plt.ylabel('Incoming radiation [W/m2]')
```

```python
# Daily energy
tau_sunset = np.arccos(-np.tan(declination_radians) * np.tan(lon))
H_d0h = 24 * 3600 / np.pi * S_0d * (tau_sunset * np.sin(lon) * np.sin(declination_radians) 
                                    + np.cos(lon) * np.cos(declination_radians) * np.sin(tau_sunset))
plt.plot(H_d0h)
```
